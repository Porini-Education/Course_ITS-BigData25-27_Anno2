```python
with open('data/riposo.csv') as f:
    file_content = f.read()
print(file_content)
```

```python
# opzione diretta per leggere e stampare le righe di un file CSV
with open('data/riposo.csv') as f:
    for line in f:
        print(line)
```

```python
header = []
data = []
for line in file_content.splitlines():
    if not header:
        header = line.split(',')
    else:
        # data.append(line.split(','))
        # non basta inserire i dati, dobbiamo sistemare la colonna della date (attualmente un range)
        line_data = line.split(',')

        # prendo la colonna della data (attualmente un range) e splitto per ottenere i valori di inizio e fine
        date_values = line_data[0].split('-')
        # print(data)

        correct_data = []
        if len(line_data) > len(header): #se il numero di dati nella riga è maggiore del numero di colonne dell'header
            # estraggo l'anno dalla colonna della data (attualmente un range)
            year = line_data[1].split('-')[0]
            correct_data.append(year)
        else:
            correct_data.append(None)

        # aggiungo il valore di "data inzio" al set di dati
        correct_data.append(date_values[0])

        start_index = len(line_data)-len(header)+1 # calcola l'indice di partenza per i dati corretti
        #aggiungo gli altri dati a partire dall'indice calcolato
        correct_data.extend(line_data[start_index:])

        data.append(correct_data) # aggiungo la riga corretta ai dati

header = ['year', 'start_date'] + header[1:]
print(len(header), len(data[0]))
```

```python
df_riposo = pd.DataFrame(data, columns=header)
df_riposo
```

```python
df_riposo.info()
```

```python
# non posso usare questa per convertire in int, perché i NaN causerebbero un errore
# df_riposo['year'] = df_riposo['year'].astype(int)
```

```python
year_fill = int(df_riposo['year'].max())+1
year_fill
```

```python
#alternativa con errors='coerce' per convertire in int ignorando i NaN
year_fill_alt = pd.to_numeric(df_riposo['year'], errors='coerce').max()
year_fill_alt
```

```python
df_riposo['year'] = df_riposo['year'].fillna(year_fill).astype(int)
df_riposo.info()
```

```python
dates = df_riposo['start_date'].str.split()
# converto il primo elemento da abbreviazione di mese a numero
mesi = ('Gen', 'Feb', 'Mar', 'Apr', 'Mag', 'Giu', 'Lug', 'Ago', 'Set', 'Ott', 'Nov', 'Dic')

nuove_date = []
for year, date in zip(df_riposo['year'], dates):
    mese_num = mesi.index(date[0]) + 1 # converto l'abbreviazione del mese in numero (+1 rispetto all'indice)
    nuove_date.append(f"{year}-{mese_num:02d}-{date[1]}") #formato YYYY-MM-DD

df_riposo['start_date'] = nuove_date
df_riposo.drop(columns=['year'], inplace=True)
```

```python
df_riposo.to_csv('data/df_riposo.csv', index=False)
df_riposo.info()
```
